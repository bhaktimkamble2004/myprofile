Google Earth Engine is a cloud-based platform for planetary-scale geospatial analysis. It integrates a vast catalog of satellite imagery and datasets with powerful computation capabilities. Users leverage it for tasks like environmental monitoring and land cover mapping.[3]:

## Key Features
The platform provides access to petabytes of remote sensing data, including Landsat and Sentinel imagery spanning over 40 years. It supports JavaScript and Python APIs via a web-based Code Editor for processing rasters and vectors at scale. Google's infrastructure enables fast, parallel computations without local downloads.[1][2][4][3]

## Common Applications
Researchers apply it to detect deforestation, track vegetation changes, and map urban expansion. It excels in time-series analysis for climate trends and disaster assessment, such as flood mapping. Non-profits and scientists use it for global-scale studies free for research.[5][8][1][3]

```python
ee.Authenticate()
ee.Initialize(project='bhakti-ee')
```

```python
m = geemap.Map(ee_initialize=False)
m
```

```python
State = ee.FeatureCollection('projects/bhakti-ee/assets/state_boundary')
m.addLayer(State, {}, 'India_States')
m
```

```python
ASSAM = State.filter(ee.Filter.eq('STATE', 'ASSAM'))
m.addLayer(ASSAM, {}, 'ASSAM')
m
```

```python
DEM = ee.Image('CGIAR/SRTM90_V4')
ASSAM_DEM = DEM.clip(ASSAM)
m.addLayer(ASSAM_DEM, {}, 'DEM')
m
```

```python
NDVI = ee.ImageCollection('LANDSAT/COMPOSITES/C02/T1_L2_8DAY_NDVI')
ASSAM_NDVI = NDVI.filterDate('2025-01-01', '2025-12-31').mosaic().clip(ASSAM)
m.addLayer(ASSAM_NDVI, {}, 'NDVI')
m
```
