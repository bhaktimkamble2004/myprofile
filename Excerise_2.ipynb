```python
import ee
```

```python
import geemap
```

```python
ee.Authenticate()
```

```python
ee.Initialize(project= "bhakti-ee")
```

```python
map = geemap.Map()
map
```

```python
SUBDISTRICT_BOUNDARY = ee.FeatureCollection("projects/bhakti-ee/assets/SUBDISTRICT_BOUNDARY")
MUMBAI = SUBDISTRICT_BOUNDARY.filter(ee.Filter.eq("Sub_dist", "MUMBAI"))
map.addLayer(MUMBAI, {}, "MUMBAI")
map.centerObject(MUMBAI, 12)
map
```

```python
DEM = ee.ImageCollection("JAXA/ALOS/AW3D30/V4_1").select("DSM")
MUMBAI_DEM = DEM.filterBounds(MUMBAI).mosaic().clip(MUMBAI)
map.addLayer(MUMBAI_DEM, {"min":0,"max":54,"palette":["#EBDABD", "#CB9960", "#D2B48C", "#B27644", "#A0522D", "#8E4F31", "#8B4513", "#7B3F00", "#663524", "#4E1F00"]}, "DEM")
map
```

```python
DEM = ee.ImageCollection("LANDSAT/COMPOSITES/C02/T1_L2_8DAY_NDVI").select("NDVI")
NDVI_1990 = DEM.filterDate("1990-01-01", "1990-12-31").filterBounds(MUMBAI).mosaic().clip(MUMBAI)
NDVI_2025 = DEM.filterDate("2025-01-01", "2025-12-31").filterBounds(MUMBAI).mosaic().clip(MUMBAI)
palette_1990 = {"min":-0.4,"max":0.5,"palette":['#E6FFE6', '#B8FFB8', '#8AFF8A', '#5CFF5C', '#2EFF2E', '#00FF00', '#00D100', '#00A300', '#008000', '#004700']}
palette_2025 = {"min":-0.6,"max":0.7,"palette":['#E6FFE6', '#B8FFB8', '#8AFF8A', '#5CFF5C', '#2EFF2E', '#00FF00', '#00D100', '#00A300', '#008000', '#004700']}
map.addLayer(NDVI_1990, palette_1990, "NDVI_1990")
map.addLayer(NDVI_2025, palette_2025, "NDVI_2025")
map

```
